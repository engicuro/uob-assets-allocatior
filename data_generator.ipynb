```python
#open_data_df
def correlation(X,Y):
    if len(X)==len(Y):
        Sum_xy = sum((X-np.mean(X))*(Y-np.mean(Y)))
        Sum_x_squared = sum((X-np.mean(X))**2)
        Sum_y_squared = sum((Y-np.mean(Y))**2)       
        corr = Sum_xy / np.sqrt(Sum_x_squared * Sum_y_squared)
    return float(corr)

from vnstock import Quote
vnind = Quote(symbol='VNINDEX', source='VCI')
vnindex_df = vnind.history(start='2019-01-01', interval='1W')
vnindex_df = vnindex_df[['time','close']]
vnindex_df['time'] = vnindex_df['time'].apply(lambda x: x.strftime('%Y-%m-%d'))
vnindex_df=vnindex_df.ffill(axis=0)


timerange = vnindex_df['time']
#corr_df.columns = ['time']
#fpt_close = fpt_close.ffill(axis=0)

backtime = 52
tmp_lst = []
for i2 in pool_lst_v2:
    tmp_lst_backtime_vnd = []
    for i0 in range(len(timerange)):
        tmp_add =[]
        for j0 in range(backtime):
            tmp_add.append(float(vnindex_df['close'].values[i0-j0]))
        tmp_lst_backtime_vnd.append(tmp_add)
    tmp_lst_backtime_vnd=tmp_lst_backtime_vnd[backtime:]
    
    stock_df = open_data_df[['time',i2]]
    tmp_lst_backtime_stock = []
    for i0 in range(len(timerange)):
        tmp_add =[]
        for j0 in range(backtime):
            tmp_add.append(float(stock_df[i2].values[i0-j0]))
        tmp_lst_backtime_stock.append(tmp_add)
    tmp_lst_backtime_stock=tmp_lst_backtime_stock[backtime:]

    corr_lst = []
    for i1 in range(len(timerange)-backtime):
        corr_lst.append(correlation(tmp_lst_backtime_vnd[i1],tmp_lst_backtime_stock[i1]))
    tmp_lst.append(corr_lst)
```

```python
corr_df = pd.DataFrame(data=tmp_lst)
corr_df=corr_df.T

columns_names = list(open_data_df.columns.values[1:])
corr_df.columns = columns_names
timerange_corr = list(timerange[backtime:])
corr_df['time'] = timerange_corr
corr_df = corr_df[['time'] + [col for col in corr_df.columns if col != 'time']]
corr_df
```

```python
corr_df.to_csv("execution_data/corr_vnindex_52w_plus1.csv")
```
