```python
for i in pool_lst_v2:
    try:
        symbol = i

        stock = Vnstock().stock(symbol=symbol, source='VCI')
        pricevolume = stock.quote.history(start='2020-01-01', interval='1W') #2020-01-01
        pricevolume['time'] = pricevolume['time'].apply(lambda x: x.strftime('%Y-%m-%d'))
        pricevolume.columns = ['time', 'open', 'high', 'low','close','volume']
        pricevolume.to_csv('raw_data/pv/'+symbol+".csv", index=False)
        time.sleep(3)
        print("mined at ",symbol)
    except:
        print("error at:",symbol)
```

```python
for i in pool_lst_v2:
    try:
        symbol = i

        stock = Vnstock().stock(symbol=symbol, source='VCI')
        financialratiodata = stock.finance.ratio(period='quarter', lang='en', dropna=True)
        #financialratiodata1= financialratiodata[[('Meta', 'ticker') , ('Meta', 'yearReport') ,('Meta', 'lengthReport') , ('Chỉ tiêu khả năng sinh lợi', 'ROE (%)') , ('Chỉ tiêu định giá', 'EV/EBITDA') , ('Chỉ tiêu định giá', 'EPS (VND)') , ('Chỉ tiêu định giá', 'P/E'),('Chỉ tiêu định giá', 'P/B')]]
        financialratiodata1= financialratiodata[[('Meta', 'ticker') , ('Meta', 'yearReport') ,('Meta', 'lengthReport') , ('Chỉ tiêu khả năng sinh lợi', 'ROE (%)')  , ('Chỉ tiêu định giá', 'EPS (VND)') , ('Chỉ tiêu định giá', 'P/E'),('Chỉ tiêu định giá', 'P/B')]]

        #financialratiodata1.columns = ['symbol', 'year', 'quarter', 'roe','ev/ebitda','eps','p/e','p/b']
        financialratiodata1.columns = ['symbol', 'year', 'quarter', 'roe','eps','p/e','p/b']
        financialratiodata1.to_csv('raw_data/financeratio/'+symbol+".csv", index=False)
        print("mined at",i)
        time.sleep(3)
    except:
        print("error at ",symbol)
```

```python
for i in pool_lst_v2:
        symbol = i
        stock = Vnstock().stock(symbol=symbol, source='VCI')
        incomestatementdata = stock.finance.income_statement(period='quarter', lang='en', dropna=True)
        incomestatementdata = incomestatementdata[['ticker','yearReport','lengthReport',"Revenue (Bn. VND)"]]
        incomestatementdata.columns = ['symbol', 'year', 'quarter', 'revenue']
        incomestatementdata.to_csv('raw_data/incomestatement/'+symbol+".csv", index=False)
        print("mined at",i)
        time.sleep(3)
```

```python
for i in pool_lst_v2:
        symbol = i

        stock = Vnstock().stock(symbol=symbol, source='VCI')
        balancesheetdata = stock.finance.balance_sheet(period='quarter', lang='en', dropna=True)
        balancesheetdata = balancesheetdata[['ticker','yearReport','lengthReport','TOTAL ASSETS (Bn. VND)','LIABILITIES (Bn. VND)',"OWNER'S EQUITY(Bn.VND)",'Cash and cash equivalents (Bn. VND)']]
        balancesheetdata.columns = ['symbol', 'year', 'quarter', 'assets','liabilities','equity','cash']
        balancesheetdata.to_csv('pool_data/balancesheet/'+symbol+".csv", index=False)
        print("mined at",i)
        time.sleep(3)
```
